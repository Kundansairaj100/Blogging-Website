```python
# weathering model
def temperature(a,b,c,time):
  temp = a*time*time + b*time +c
  return temp
```

```python
# CASE-1: Hardcoded Variables
a = 3.4
b = 5.6
c = 7.8
time = 10
print(f"The temperature at time: {time} is {temperature(a,b,c,time)}")
```

```python
# CASE-2: Keyboard Input
a = float(input("Enter A value: "))
b = float(input("Enter B value: "))
c = float(input("Enter C value: "))
time = int(input("Enter Time: "))
print(f"The temperature at time: {time} is {temperature(a,b,c,time)}")
```

```python
# File Reading Function
def file_read(input):
  with open('input.txt','r') as file:
    lines = file.readlines()
    input = [tuple(map(float,line.strip().split(','))) for line in lines]

  return input
```

```python
# Single File Inputs
input = file_read('input.txt')
a,b,c,time = input[0]
print(f"The Temperature is: {temperature(a,b,c,time)}")
```

```python
#Multiple File Inputs
input = file_read('input.txt')
for i in input:
  a,b,c,time = i
  print(f"The Temperature is: {temperature(a,b,c,time)}")
```
